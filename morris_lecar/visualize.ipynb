```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import matplotlib as mpl
mpl.rcParams["font.family"] = "Nimbus Roman"
```

```python
q = np.linspace(0, 500, 20)
gbar = np.linspace(0, 20, 20)

Q, Gbar = np.meshgrid(q, gbar)  # Create mesh grid
```

```python
scores25 = pd.read_csv("./results/freq_25.00_scores.csv", index_col=0)
scores25.head()
```

```python
scores_arr25 = scores25.to_numpy()
scores_arr25
```

```python
plt.pcolormesh(Q, Gbar, np.log10(scores_arr25), shading='auto', cmap='viridis')
plt.colorbar()
plt.xlabel("Q")
plt.ylabel("gbar")
plt.title("Test RMSE for 25 Hz sine wave")
plt.savefig("./img/force_sine_25_Hz.jpg", dpi=200, bbox_inches="tight")
plt.show()
```

```python
scores5 = pd.read_csv("./results/freq_5.00_scores.csv", index_col=0)
scores5.head()
```

```python
scores_arr5 = scores5.to_numpy()
scores_arr5
```

```python
plt.pcolormesh(Q, Gbar, np.log10(scores_arr5), shading='auto', cmap='viridis')
cbar = plt.colorbar()
cbar.set_label(r"$\log{RMSE}$")
plt.xlabel("Q")
plt.ylabel("gbar")
plt.title("Test RMSE for 5 Hz sine wave")
plt.savefig("./img/force_sine_5_Hz.jpg", dpi=200, bbox_inches="tight")
plt.show()
```
