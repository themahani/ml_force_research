## Import libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import torch
from tqdm import tqdm


import sys, os

from rctorch.models import MorrisLecar, MorrisLecarCurrent
from rctorch import Reservoir
from rctorch.utils import minmax_transform, z_transform
from rctorch.supervisors import LorenzAttractor

from scipy.signal import find_peaks
```

## Preparations for Lorenz Attractor

```python
seed = 1
np.random.seed(seed)
torch.cuda.manual_seed(seed)
torch.random.manual_seed(seed)

T = 20000
dt = 1e-1
t = np.arange(0, T, dt)
nt = t.size
# x = VanDerPol(T, dt, mu=1, tau=.02).generate(transient_time=500.0)
x = LorenzAttractor(T, dt, tau=.01).generate(transient_time=1000.0)

x = x.T
sup = minmax_transform(x, zero_mean=True)
# sup = z_transform(x)
print(sup.shape)
```

# ML Conductance-based Model

### Prep work

```python
NE = 400
NI = 100
N = NI + NE


# input current for I and E neurons
Ie = 70     # pA
Ii = 70     # pA
current = np.array([Ie] * NE + [Ii] * NI).reshape(N, 1)


Q = 300
delta = 1.0
gbar = 5.0
sparse = 0.01
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else 'cpu')
print(f"Using Device <{device}> for PyTorch computations...\n")
torch.cuda.random.manual_seed(seed)
res = None
res_tgn = None
try:
    res = Reservoir(    # NOTE: This is the main reservoir
        n_input=sup.shape[1], n_output=sup.shape[1], w_in_amp=0, model_cls=MorrisLecar,
        # p_sparsity=sparse,
        BIAS=current, dt=dt, 
        Ne=NE, Ni=NI, gbar=gbar, device=device,
        )
    res_tgn = Reservoir(    # NOTE: This is the reservoir used for TGN which will be discarded after training
    n_input=sup.shape[1], n_output=sup.shape[1], w_in_amp=Q, model_cls=MorrisLecar,
    # p_sparsity=sparse,
    BIAS=current, dt=dt, 
    Ne=NE, Ni=NI, gbar=gbar, device=device,
    )
    res.model.w = res.model.w.float()
    res_tgn.model.w = res.model.w.clone().detach()
except Exception as e:
    print(e)
    print(torch.cuda.memory_summary(device=device))
```

```python
# colors = plt.pcolormesh(res.w.cpu().numpy(), cmap='viridis')
# plt.colorbar(colors)
# plt.show()
factory_kwargs = res.model.factory_kwargs

print(f"Mean: {res.model.w.type(dtype=torch.float64).mean().item()}, \nSTD: {res.model.w.type(dtype=torch.float64).std().item()}")
print(f"Max: {res.model.w.type(dtype=torch.float64).max().item()}, \nMin: {res.model.w.type(dtype=torch.float64).min().item()}")
```

```python
print(res.W_in.max().item(), res.W_in.min().item())
print(res_tgn.W_in.max().item(), res_tgn.W_in.min().item())
```

```python
# RLS params
# rls_stop = round(T * .6)
rls_start = 500
test_time = T * .5    # ms
rls_stop = T - test_time
rls_step = 50

memory_length = 5000    # ms
rls_gap = rls_step * dt
memory_coeff = memory_length / rls_gap
ff_coeff = 1 - 1 / memory_coeff

ff_coeff = 1
ff_coeff
```

```python
save_path = os.path.join(os.getcwd(), "results", "conductance",
                         f"Q_{Q}_gbar_{gbar}_l_{delta}_Ne_{NE}_Ni_{NI}")
os.makedirs(save_path, exist_ok=True)

def save_tensor(tensor, name):
    file_name = name + ".pt"
    torch.save(tensor, os.path.join(save_path, file_name))
    print(f"Saved {file_name} in {save_path}")
    
def load_tensor(save_path:str, name:str):
    file_name = os.path.join(save_path, name + ".pt")
    tensor = torch.load(os.path.join(save_path, file_name))
    print(f"Loaded {file_name} from {save_path}")
    return tensor
```

```python
# Save parameters in  the save_path in JSON format
def save_params(save_path, params):
    import json
    with open(os.path.join(save_path, "params.json"), "w") as f:
        json.dump(params, f, indent=4)
    print(f"Saved parameters in {save_path}/params.json")

params = {
    "Q": Q,
    "gbar": gbar,
    "delta": delta,
    "sparse": sparse,
    "Ie": Ie,
    "Ii": Ii,
    "NE": NE,
    "NI": NI,
    "memory_length": memory_length,
    "rls_start": rls_start,
    "rls_stop": rls_stop,
    "rls_step": rls_step,
    "ff_coeff": ff_coeff,
    "T": T,
    "dt": dt,
    "seed": seed,
}

save_params(save_path, params)
```

```python
duration = 500
nt_transient = int(duration / dt)

print(f"Transient Period: {duration} ms")
for i in tqdm(range(nt_transient)):
    res.model.forward(0)
    res_tgn.model.forward(0)
```

### Training

```python
n_neurons = 10
neurons = np.random.choice(N, n_neurons, replace=False)

sup_tensor = torch.tensor(sup, **factory_kwargs)
nt_train = int((rls_stop) / dt)

xhat_rec_train = torch.zeros((nt_train, sup.shape[1]), **factory_kwargs)

delta = 5
ml = res.model
ml_tgn = res_tgn.model
Pinv = torch.eye(res.n_hidden, **factory_kwargs) / delta

print(f"Training Period: {rls_stop} ms")
for i in tqdm(range(nt_train)):
    input_fb_tgn = res_tgn.W_in @ sup_tensor[i].reshape(-1, 1)   # (n_hidden, 1)
    # print(input_fb_tgn.size())
    I_tgn = ml_tgn.calc_ipsc() + input_fb_tgn
    ml_tgn.forward(input_fb_tgn)
    x_hat = res.W_out.T @ ml.s          # (n_output, 1)
    # print(x_hat.size())
    xhat_rec_train[i] = x_hat.ravel()
    I_ml = ml.calc_ipsc()
    ml.forward(0)
    # dec_rec_train[i] = res.W_out[neurons]
    if i % rls_step == 0:
        # x = sup_tensor[i].reshape(-1, 1)    # (n_output, 1)
        # e_rec = I_ml - I_tgn                # (n_hidden, 1)
        # e_out = res.W_out.T @ ml.s - x      # (n_output, 1)
        # u = Pinv @ ml.s                     # (n_hidden, 1)
        # k = u / (ff_coeff + ml.s.t() @ u)
        # Pinv = (Pinv - k @ (u.T)) / ff_coeff
        # res.W_out -= k @ e_out.T
        # ml.w -= k @ e_rec.T         # (n_hidden, n_hidden)
        
        x = sup_tensor[i].reshape(-1, 1)    # (n_output, 1)
        e_rec = I_ml - I_tgn                # (n_hidden, 1)
        e_out = res.W_out.T @ ml.s - x      # (n_output, 1)
        Pinv -= (Pinv @ ml.s @ ml.s.T @ Pinv) / (1 + ml.s.T @ Pinv @ ml.s)
        res.W_out -= (Pinv @ ml.s) @ e_out.T
        ml.w -= (Pinv @ ml.s) @ e_rec.T         # (n_hidden, n_hidden)
```

```python
hists = []
for axis in range(sup.shape[1]):
    sup_hist, bins = np.histogram(sup[-nt_test:, axis], bins=200, density=True)
    output_hist, bins = np.histogram(x_hat_rec[:, axis].cpu().numpy(), bins=bins, density=True)
    hists.append((sup_hist, output_hist, bins))

# KL Divergence
kl_divs = []
for i in range(len(hists)):
    sup_hist, output_hist, _ = hists[i]
    sup_hist = sup_hist / np.sum(sup_hist)  # normalize
    output_hist = output_hist / np.sum(output_hist) # normalize
    # Mask zeros
    sup_hist = np.where(sup_hist == 0, 1e-10, sup_hist)
    output_hist = np.where(output_hist == 0, 1e-10, output_hist)
    # KL Divergence
    kl_div = np.sum(sup_hist * np.log(sup_hist / output_hist))
    print(f"KL Divergence for axis {i}: {kl_div}")
    kl_divs.append(kl_div)
```

```python
fig, ax = plt.subplots(figsize=(20, 5), ncols=len(hists), sharey=True)

ax_names = ['x', 'y', 'z']

for axis in range(len(ax)):
    bins = hists[axis][2]
    sup_hist = hists[axis][0]
    output_hist = hists[axis][1]
    # bins = (bins[:-1] + bins[1:]) / 2   # center of the bins
    width = bins[1] - bins[0]
    alpha = .5
    ax[axis].bar(bins[:-1], sup_hist, width=width, label="supervisor", color='b', alpha=alpha)
    ax[axis].bar(bins[:-1], output_hist, width=width, label="output", color='r', alpha=alpha)
    ax[axis].set_xlabel(f"{ax_names[axis]}", fontsize=16)
    ax[axis].legend(loc='upper right')
    ax[axis].set_title(f"KL Divergence: {kl_divs[axis]:.4f}", fontsize=16)
ax[0].set_ylabel("Density", fontsize=16)
plt.suptitle(f"I={Ie} pA, Q={Q}, gbar={gbar}, $\delta$={delta}", fontsize=16, y=1.05)
plt.savefig(os.path.join(save_path, f"N_{ml.N}_ff_{ff_coeff}_histogram.svg"), dpi=300, bbox_inches='tight')
plt.show()
```
